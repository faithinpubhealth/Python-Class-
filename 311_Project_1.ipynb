```python
#visualization 1 - Number of complaints based on city agency - we see that most complaints go to NYPD
df.groupby('Agency')["Unique Key"].count().sort_values(ascending=False).plot(kind="bar", title = "Distribution of Complaints to each City Agency", ylabel= "number of complaints", xlabel = "City Agency", color = 'purple')
```

```python
#which borough has most complaints?  Brooklyn has the most complaints
df.groupby('Borough')['Unique Key'].count().sort_values(ascending=False)
```

```python
#Table of count of Noise complaints by Borough
df[df['Complaint Type']== "Noise"].groupby('Borough')['Unique Key'].count().sort_values(ascending=False)
```

```python
#PLot by count for noise complain"ts
df[df['Complaint Type']== "Noise"].groupby('Borough')['Unique Key'].count().sort_values(ascending=False).plot(kind="bar", title="Count of Noise Complaints by Borough", color = "salmon")
```

```python
#Find one more intersting information
#Top 5 complaints in (specific zip code) Home Zip code
df[df['Incident Zip']== "11226"].groupby('Complaint Type')['Unique Key'].count().sort_values(ascending=False).head(5)
```
